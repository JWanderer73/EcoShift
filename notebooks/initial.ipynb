```python
import requests
import pandas as pd

url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude": 32.7157,
    "longitude": -117.1611,

    "start_date": "2026-01-01",
    "end_date": "2026-10-20",

    "hourly": [
        "temperature_2m",
        "relative_humidity_2m", # How humid the air is
        "wind_speed_10m", # Wind speed
        "wind_direction_10m",
        "shortwave_radiation",
        "direct_radiation",
        "diffuse_radiation",
        "cloud_cover",
        "precipitation",
        "is_day",
    ],

    "timezone": "UTC",
}

response = requests.get(url, params=params)
response.raise_for_status()  # Raise an error if the request was unsuccessful

data = response.json()
df = pd.DataFrame(data["hourly"])

df['time'] = pd.to_datetime(df['time'],utc=True)

display(df)
```

```python
# Check the dataset dimensions
print("Shape:", df.shape)

# Check missing values in each column
print("\nMissing values:")
print(df.isna().sum())

# Inspect the last 10 rows
display(df.tail(10))

# Check whether timestamps are duplicated
print(
    "\nDuplicate timestamps:",
    df["time"].duplicated().sum()
)

# Check the first and last timestamps
print("\nFirst timestamp:", df["time"].min())
print("Last timestamp:", df["time"].max())
```

```python
df.to_csv("open_meteo_weather_2026.csv", index=False)
```
